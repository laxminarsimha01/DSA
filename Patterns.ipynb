#### Pattern 1: Rectangular Star pattern

```python
def RectangularStar(n):
    for i in range(n):
       for j in range(n):
           print("*", end=' ')
       print()

RectangularStar(5) 
```

#### Pattern 2: Right angled Triangle

```python
def RightAngled1(size):
    for i in range(size+1):
        for j in range(i):
            print('*', end=' ')
        print()
        
RightAngled1(5)
```

#### Pattern 3: Right Angled Triangle with sequence

```python
def RightAngled2(size):
    for i in range(size+1):
        for j in range(1,i+1):
            print(j, end=' ')
        print()
RightAngled2(5)
```

#### Pattern 4: Right Angled Traingle repeating number

```python
def RightAngled3(size):
    for i in range(size+1):
        for j in range(1, i+1):
            print(i, end=' ')
        print()
RightAngled3(5)
```

#### Pattern 5: Reverse Right angled triangle

```python
def ReverseRightAngled(size):
    for i in range(size):
        for j in range(size-i):
            print('*', end=' ')
        print()
ReverseRightAngled(5)
```

#### Pattern 6: Reverse Right Angled Triangle Sequence 

```python
def ReverseRightAngled2(size):
    for i in range(size+1):
        for j in range(1,size-i+1):
            print(j, end=' ')
        print()
ReverseRightAngled2(5)
```

#### Pattern 7: Equilateral Triangle

```python
def EquiTriangle(size):
    for i in range(size):
        for j in range(size-i): #Space
            print(" ", end=' ')
        for x in range(2*i+1):  #Star
            print("*", end=' ')
        print()
EquiTriangle(5)
```

#### Pattern 8: Reverse Equilateral Triangle

```python
def ReverseEquiTriangle(size):
    for i in range(size):
        for j in range(i): #Space
            print(' ', end=' ')
        for x in range(2*(size-i)-1):
            print('*', end=' ')
        print()
ReverseEquiTriangle(5)
```
