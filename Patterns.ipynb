#### Pattern 1: Rectangular Star pattern

```python
def RectangularStar(n):
    for i in range(n):
       for j in range(n):
           print("*", end=' ')
       print()

RectangularStar(5) 
```

#### Pattern 2: Right angled Triangle

```python
def RightAngled1(size):
    for i in range(size+1):
        for j in range(i):
            print('*', end=' ')
        print()
        
RightAngled1(5)
```

#### Pattern 3: Right Angled Triangle with sequence

```python
def RightAngled2(size):
    for i in range(size+1):
        for j in range(1,i+1):
            print(j, end=' ')
        print()
RightAngled2(5)
```

#### Pattern 4: Right Angled Traingle repeating number

```python
def RightAngled3(size):
    for i in range(size+1):
        for j in range(1, i+1):
            print(i, end=' ')
        print()
RightAngled3(5)
```

#### Pattern 5: Reverse Right angled triangle

```python
def ReverseRightAngled(size):
    for i in range(size):
        for j in range(size-i):
            print('*', end=' ')
        print()
ReverseRightAngled(5)
```

#### Pattern 6: Reverse Right Angled Triangle Sequence 

```python
def ReverseRightAngled2(size):
    for i in range(size+1):
        for j in range(1,size-i+1):
            print(j, end=' ')
        print()
ReverseRightAngled2(5)
```

#### Pattern 7: Equilateral Triangle

```python
def EquiTriangle(size):
    for i in range(size):
        for j in range(size-i): #Space
            print(" ", end=' ')
        for x in range(2*i+1):  #Star
            print("*", end=' ')
        print()
EquiTriangle(5)
```

#### Pattern 8: Reverse Equilateral Triangle

```python
def ReverseEquiTriangle(size):
    for i in range(size):
        for j in range(i): #Space
            print(' ', end=' ')
        for x in range(2*(size-i)-1):
            print('*', end=' ')
        print()
ReverseEquiTriangle(5)
```

#### Pattern 9: Rhombus

```python
def rhombus(n):
    for i in range(n):
        print(' '*(n-i-1)+'*'*(2*i+1))
    for j in range(n):
        print(' '*j+'*'*(2*(n-j)-1))
rhombus(5)
```

#### Pattern 10: Semi Rhombus

```python
def SemiRhombus(n):
    for i in range(n):
        print('*'*(i+1))
    for j in range(n-1):
        print('*'*(n-j-1))
SemiRhombus(5)
```

#### Pattern 11: Binary Right angled triangle

```python
def BinaryRight(n):
    for i in range(n+1):
        if i%2==0:
            for j in range(i):
                print(1, end=' ') if j%2!=0 else print(0, end=' ')
            print()
        else:
            for x in range(i):
                print(0, end=' ') if x%2!=0 else print(1, end=' ')
            print()
BinaryRight(5)
```

#### Pattern 12: Crown Number Pattern

```python
def CrownNumber(n):
    for i in range(1, n+1):
        for j in range(1, i+1):
            print(j, end=' ')
        print(' '*(2*(n+1)-2*i), end=' ')
        while(i!=0):
            print(i, end=' ')
            i-=1
        print()
CrownNumber(5)
```
